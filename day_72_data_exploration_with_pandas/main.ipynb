```python
# What college major has the highest mid-career salary? How much do graduates with this major earn?
idx = clean_df['Mid-Career Median Salary'].idxmax()
print(clean_df.loc[idx]['Mid-Career Median Salary'])
print(f"Index for the max mid career salary: {idx}")
clean_df['Undergraduate Major'][idx]
```

```python
# Which college major has the lowest starting salary and how much do graduates earn after university?
idx = clean_df['Starting Median Salary'].idxmin()
print('Major:', clean_df.loc[idx]['Undergraduate Major'])
print('Major:', clean_df.loc[idx]['Starting Median Salary'])
```

```python
# Which college major has the lowest mid-career salary and how much can people expect to earn with this degree? 
idx = clean_df['Mid-Career Median Salary'].idxmin()
print('Major:', clean_df.loc[idx]['Undergraduate Major'])
print('Major:', clean_df.loc[idx]['Starting Median Salary'])
```

```python
clean_df['Mid-Career 90th Percentile Salary'] - clean_df['Mid-Career 10th Percentile Salary']
```

```python
spread_col = clean_df['Mid-Career 90th Percentile Salary'] - clean_df['Mid-Career 10th Percentile Salary']
clean_df.insert(1, 'Spread', spread_col)
clean_df.head()
```

```python
low_risk = clean_df.sort_values('Spread', ascending=False)
low_risk[['Undergraduate Major', 'Spread']].head()
```

```python
# Using the .sort_values() method, can you find the degrees with the highest potential? 
# Find the top 5 degrees with the highest values in the 90th percentile. 

high_potential = clean_df.sort_values('Mid-Career 90th Percentile Salary', ascending=False)
high_potential[['Undergraduate Major', 'Mid-Career 90th Percentile Salary']].head()
```

```python
# Find the degrees with the greatest spread in salaries. 
# Which majors have the largest difference between high and low earners after graduation.   

great_spread = clean_df.sort_values('Spread', ascending=False)
great_spread[['Undergraduate Major','Spread']].head()
```

```python

great_spread = clean_df.sort_values('Mid-Career Median Salary', ascending=False)
great_spread[['Undergraduate Major','Mid-Career Median Salary']].head()
```
